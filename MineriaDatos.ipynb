#####IPython Magic Commands

```python
#para evaluar el desempeño en base al tiempo
%timeit L = [n ** 2 for n in range(1000)]
```

```python
#con doble %% podemos evakluar el desmpeño en cuanto al tiempo; en mas de una linea de codigo
%%timeit
L = []
for n in range(1000):
  L.append(n ** 2)
```

#####Understanding Data Types in Python

```python
#importa numpy y renombralo como np
#libreria para arreglos de n dimenciones
import numpy as np
```

```python
#arreglo lineal de enteros
np.array([1, 4, 2, 5, 3])
```

```python
#arreglo donde hace un cast a flotante
#el primer dato es float
np.array([3.14, 4, 2, 3])
```

```python
#cast del array a float
np.array([1, 2, 3, 4], dtype='float32')
```

```python
#cast del array a entero
np.array([1, 2, 3, 4], dtype='int')
```

```python
#cuando el argumento de np.array es una iteracion anidada, genera una matriz
np.array([range(i, i + 3) for i in [2, 4, 6]])
```
