#####Understanding Data Types in Python

```python
#importa numpy y renombralo como np
#libreria para arreglos de n dimenciones
import numpy as np
```

```python
#arreglo lineal de enteros
np.array([1, 4, 2, 5, 3])
```

```python
#arreglo donde hace un cast a flotante
#el primer dato es float
np.array([3.14, 4, 2, 3])
```

```python
#cast del array a float
np.array([1, 2, 3, 4], dtype='float32')
```

```python
#cast del array a entero
np.array([1, 2, 3, 4], dtype='int')
```

```python
#cuando el argumento de np.array es una iteracion anidada, genera una matriz
np.array([range(i, i + 3) for i in [2, 4, 6]])
```

######Creating Arrays from Scratch

```python
# Create a length-10 integer array filled with zeros
np.zeros(10, dtype=int)
```

```python
# Create a 3x5 floating-point array filled with ones
np.ones((3, 5), dtype=float)
```

```python
# Create a 3x5 array filled with 3.14
np.full((3, 5), 3.14)
```

```python
# Create an array filled with a linear sequence
# Starting at 0, ending at 20, stepping by 2
# (this is similar to the built-in range() function)
#comienza en 0, llega hasta antes de 20 y se incrementa de dos en dos
np.arange(0, 20, 2)
```

```python
# Create an array of five values evenly spaced between 0 and 1
np.linspace(0, 1, 5)
```

```python
# Create a 3x3 array of uniformly distributed
# random values between 0 and 1
np.random.random((3, 3))
```

```python
# Create a 3x3 array of normally distributed random values
# with mean 0 and standard deviation 1
#.normal(media, desviacion estandar, (matriz de x, x))
np.random.normal(0, 1, (3, 3))
```

```python
# Create a 3x3 array of random integers in the interval [0, 10)
np.random.randint(0, 10, (3, 3))
```

```python
# Create a 3x3 identity matrix
#cuando la diagonal principal tiene puros 1´s
np.eye(3)
```

```python
# Create an uninitialized array of three integers
# The values will be whatever happens to already exist at that memory location
np.empty(3)
```

#####The Basics of NumPy Arrays

######NumPy Array Attributes

```python
np.random.seed(0)  # seed for reproducibility

x1 = np.random.randint(10, size=6)  # One-dimensional array
x2 = np.random.randint(10, size=(3, 4))  # Two-dimensional array
x3 = np.random.randint(10, size=(3, 4, 5))  # Three-dimensional array
```

```python
print("x3 ndim: ", x3.ndim)
print("x3 shape:", x3.shape)
print("x3 size: ", x3.size)
```

```python
print("dtype:", x3.dtype)
```

```python
print("itemsize:", x3.itemsize, "bytes")
print("nbytes:", x3.nbytes, "bytes")
```

####Análisis Estadístico Básico con NumPy

#####Medidas Estadísticas Básicas

######Media (Mean)

```python
data = np.array([1, 2, 3, 4, 5])
media = np.mean(data)
print(media)
```

######Promedio (Average) y Promedio Ponderado

```python
data = np.array([1, 2, 3, 4, 5])
weights = np.array([1, 2, 3, 4, 5])
promedio = np.average(data)
promedio_ponderado = np.average(data, weights=weights)
```

######Mediana

```python
mediana = np.median(data)
```

######Varianza y Desviación Estándar

```python
varianza = np.var(data)
desviacion = np.std(data)
```

#####Generación de Datos Aleatorios

```python
muestra = np.random.normal(0, 1, 1000)
print(np.mean(muestra), np.std(muestra))
```
