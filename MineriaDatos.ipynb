```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import drive

# 1. MONTAR GOOGLE DRIVE
# Al ejecutar esto, te pedirá permiso para acceder a tus archivos.
drive.mount('/content/drive')

# 2. CARGAR Y LIMPIAR DATOS
# Asumimos que el archivo está en la raíz de "Mi unidad" (MyDrive).
# Si está en una carpeta, cambia la ruta (ej: '/content/drive/MyDrive/Carpetas/CUADROS_ESTADISTICOS.CSV')
ruta_archivo = '/content/drive/MyDrive/CUADROS_ESTADISTICOS.CSV'

try:
    df = pd.read_csv(ruta_archivo)

    # --- BLOQUE DE LIMPIEZA ---
    # Tu archivo CSV parece tener una 't' extra al inicio de las columnas y valores.
    # Esto las elimina para que coincidan con tu lógica original ("tSi" -> "Si").

    # Limpiar nombres de columnas (quitar espacios y 't' inicial)
    df.columns = [c.strip().lstrip('t') for c in df.columns]

    # Limpiar valores de texto (quitar 't' inicial de las respuestas)
    columnas_texto = df.select_dtypes(include=['object']).columns
    for col in columnas_texto:
        # Convertimos a string, quitamos espacios y la 't' de la izquierda
        df[col] = df[col].astype(str).str.strip().str.lstrip('t')
    # --- FIN LIMPIEZA ---

except FileNotFoundError:
    print("¡Error! No se encontró el archivo. Verifica que el nombre y la ruta sean correctos en tu Drive.")
    raise

# 3. PROCESAMIENTO (Tu lógica original intacta)

# Variables
# Convertimos edad a numérico (ahora que ya limpiamos la 't', funcionará correctamente)
df['edad'] = pd.to_numeric(df['¿Cuántos años cumplidos tienes?'], errors='coerce')

map_sn = {'Si': True, 'No': False}
df['prog_python'] = df['¿Has realizado programas utilizando Python?'].map(map_sn)
df['usa_colab'] = df['¿Has utilizado Google Colab?'].map(map_sn)
df['cuenta_github'] = df['¿Tienes cuenta de GitHub?'].map(map_sn)

# A NumPy
edad = df['edad'].dropna().to_numpy(dtype=float) # Agregué dropna por seguridad
prog_py = df['prog_python'].dropna().to_numpy(dtype=bool)
usa_colab = df['usa_colab'].dropna().to_numpy(dtype=bool)
cuenta_gh = df['cuenta_github'].dropna().to_numpy(dtype=bool)

# Aseguramos que los arrays tengan el mismo tamaño tras limpiar nulos si es necesario
# Para este ejemplo simple asumimos datos completos, pero en producción es ideal alinear los índices.

# Descriptivos de edad (NumPy)
summary_age = {
    'n': int(edad.size),
    'min': float(np.min(edad)),
    'p25': float(np.percentile(edad, 25)),
    'mediana': float(np.median(edad)),
    'media': float(np.mean(edad)),
    'p75': float(np.percentile(edad, 75)),
    'max': float(np.max(edad)),
    'rango(ptp)': float(np.ptp(edad)),
    'var_muestral': float(np.var(edad, ddof=1)),
    'std_muestral': float(np.std(edad, ddof=1)),
    'cv': float(np.std(edad, ddof=1)/np.mean(edad)) if np.mean(edad) != 0 else 0,
}

# Frecuencias (NumPy)
# Nota: Usamos len(df) para mantener la consistencia con el total de filas cargadas
total_rows = len(df)
freq_python = {'Si': int(np.count_nonzero(prog_py)), 'No': int(total_rows - np.count_nonzero(prog_py))}
freq_colab  = {'Si': int(np.count_nonzero(usa_colab)), 'No': int(total_rows - np.count_nonzero(usa_colab))}
freq_gh     = {'Si': int(np.count_nonzero(cuenta_gh)), 'No': int(total_rows - np.count_nonzero(cuenta_gh))}

# Tablas cruzadas (NumPy)
def tabla2x2(a, b):
    # Aseguramos que los arrays sean del mismo tamaño cortando al mínimo común (por seguridad si hubo nulos)
    min_len = min(len(a), len(b))
    a = a[:min_len]
    b = b[:min_len]

    t = np.zeros((2,2), dtype=int)
    t[1,1] = np.count_nonzero(a & b)
    t[1,0] = np.count_nonzero(a & ~b)
    t[0,1] = np.count_nonzero(~a & b)
    t[0,0] = np.count_nonzero(~a & ~b)
    return t

tab_py_colab = tabla2x2(prog_py, usa_colab)
tab_py_gh    = tabla2x2(prog_py, cuenta_gh)
tab_colab_gh = tabla2x2(usa_colab, cuenta_gh)

# Medias de edad por grupo (NumPy)
# Nota: Aquí debemos usar booleanos alineados con la variable edad.
# Usaremos las columnas originales del DF para asegurar alineación de índices.
means_by_group = {
    'edad_promedio_Python': {
        'Si': float(df[df['prog_python'] == True]['edad'].mean()),
        'No': float(df[df['prog_python'] == False]['edad'].mean())
    },
    'edad_promedio_Colab':  {
        'Si': float(df[df['usa_colab'] == True]['edad'].mean()),
        'No': float(df[df['usa_colab'] == False]['edad'].mean())
    },
    'edad_promedio_GitHub': {
        'Si': float(df[df['cuenta_github'] == True]['edad'].mean()),
        'No': float(df[df['cuenta_github'] == False]['edad'].mean())
    },
}

# Imprimir resultados
print("--- Resumen Edad ---")
print(summary_age)
print("\n--- Frecuencias ---")
print(f"Python: {freq_python}")
print(f"Colab:  {freq_colab}")
print(f"GitHub: {freq_gh}")
print("\n--- Tablas Cruzadas ---")
print("Python vs Colab:\n", tab_py_colab)
print("\n--- Medias por Grupo ---")
print(means_by_group)
```

Primer intento de modelar los años de escolaridad y el ingreso estimado.
Utilizar el primer punto y el útlimo para trazar una recta:

$$ m = \frac{y_{2}-y_{1}} {x_{2}-x_{1}} $$

$$ b = y_{1} - m*x_{1} $$


```python
x1 = anios_escolaridad[0]
y1 = ingreso_mensual_estimado_mx[0]
x2 = anios_escolaridad[-1]
y2 = ingreso_mensual_estimado_mx[-1]

m = (y2 - y1) / (x2 - x1)
b = y1 - m * x1

print(f"La ecuación de la recta es: y = {m:.2f}x + {b:.2f}")
```

```python
import matplotlib.pyplot as plt

# Calcular los valores y de la recta utilizando m y b
linea_ingreso_estimado_new_plot = m * anios_escolaridad + b

plt.figure(figsize=(10, 6))
plt.scatter(anios_escolaridad, ingreso_mensual_estimado_mx, label='Datos Originales')
plt.plot(anios_escolaridad, linea_ingreso_estimado_new_plot, color='red',
         label=f'Recta de Regresión (y = {m:.2f}x + {b:.2f})')
plt.title('Años de Escolaridad vs. Ingreso Mensual Estimado con Línea de Regresión')
plt.xlabel('Años de Escolaridad')
plt.ylabel('Ingreso Mensual Estimado (MXN)')
plt.grid(True)
plt.legend()
plt.show()
```

Valores predichos
: Son los ingresos mensuales que nuestra recta de regresión predice para cada año de escolaridad.

Errores (residuales): Es la diferencia entre el ingreso mensual real (ingreso_mensual_estimado_mx) y el ingreso mensual predicho (y_pred). Un error positivo significa que la predicción fue menor que el valor real, y un error negativo indica que la predicción fue mayor.

Errores al cuadrado: Hemos elevado al cuadrado cada uno de los errores. Esto se hace para eliminar los signos negativos (de modo que los errores positivos y negativos no se cancelen entre sí) y para penalizar más los errores grandes.

Suma de errores cuadrados: Es la suma de todos los errores al cuadrado. Este valor nos da una medida total de la 'distancia' de nuestros puntos a la línea.

Error Cuadrático Medio (MSE): Es la suma de los errores cuadrados dividida por el número total de puntos de datos. El MSE es una medida promedio de la magnitud de los errores. Un MSE más bajo indica que el modelo se ajusta mejor a los datos.

En este caso, el Error Cuadrático Medio (MSE) es de aproximadamente 13,196,667.31. Este número nos da una idea del rendimiento general de la recta que trazamos, mostrando la magnitud promedio de los errores al cuadrado entre los valores observados y los valores predichos por la línea.

Fórmula del MSE (Error Cuadrático Medio) del Inglés Mean Squared Error

$$ MSE = \frac{\Sigma_{i=1}^{n}(Y_{1}-\hat{Y_{i}})^{2}} {n} $$


Donde:

$MSE = $error cuadrático medio.

$n = $ número de datos.

$Y_{1} = $ valores observados.

$\hat{Y_{i}} = $ valores predichos.

```python
# 1. Calcular los valores predichos (y_pred) usando la ecuación de la recta
y_pred = m * anios_escolaridad + b
y_pred
```

```python
# 2. Calcular los errores (residuales): la diferencia entre los valores reales y los predichos
errores = ingreso_mensual_estimado_mx - y_pred
errores
```

```python

# 3. Elevar al cuadrado cada error
errores_cuadrados = errores**2
errores_cuadrados
```

```python
# 4. Sumar los errores cuadrados
suma_errores_cuadrados = np.sum(errores_cuadrados)
suma_errores_cuadrados
```

```python
# 5. Dividir la suma de los errores cuadrados por el número de puntos para obtener el MSE
mse = suma_errores_cuadrados / len(anios_escolaridad)
mse
```

###### Fórmulas de Indicadores de Evaluación

* **RMSE (Raíz del Error Cuadrático Medio):**
$$RMSE = \sqrt{MSE} = \sqrt{\frac{\sum_{i=1}^{n}(Y_{i} - \hat{Y_{i}})^2}{n}}$$

* **MAE (Error Absoluto Medio):**
$$MAE = \frac{1}{n} \sum_{i=1}^{n} |Y_{i} - \hat{Y_{i}}|$$

* **$R^2$ (Coeficiente de Determinación):**
$$R^2 = 1 - \frac{\sum_{i=1}^{n}(Y_{i} - \hat{Y_{i}})^2}{\sum_{i=1}^{n}(Y_{i} - \bar{Y})^2}$$

* **$R^2$ Ajustado:**
$$R^2_{adj} = 1 - \left( \frac{(1 - R^2)(n - 1)}{n - p - 1} \right)$$
*(Donde $n$ = número de datos, $p$ = número de variables independientes)*

* **MAPE (Error Porcentual Absoluto Medio):**
$$MAPE = \frac{100\%}{n} \sum_{i=1}^{n} \left| \frac{Y_{i} - \hat{Y_{i}}}{Y_{i}} \right|$$

#####Calcula los indicadores para la regresión lineal.

```python
from sklearn.metrics import mean_absolute_error, r2_score

mse_sk = mean_squared_error(y, y_pred_sklearn)
rmse_sk = np.sqrt(mse_sk)
mae_sk = mean_absolute_error(y, y_pred_sklearn)
r2_sk = r2_score(y, y_pred_sklearn)
n = len(y) # N observaciones
p = 1      # N variables ind
r2_adj_sk = 1 - (1 - r2_sk) * (n - 1) / (n - p - 1)
mape_sk = np.mean(np.abs((y - y_pred_sklearn) / y)) * 100

print("--- Indicadores de Regresión Lineal ---")
print(f"MSE:          {mse_sk:,.2f}")
print(f"RMSE:         {rmse_sk:,.2f} MXN")
print(f"MAE:          {mae_sk:,.2f} MXN")
print(f"R^2:          {r2_sk:.4f}")
print(f"R^2 Ajustado: {r2_adj_sk:.4f}")
print(f"MAPE:         {mape_sk:.2f}%")
```

| :--- | :---: | :---: | :---: |
| Hipertenso | 21 | 36 | 30 |

Si se selecciona aleatoriamente a uno de estos individuos, encuentre la probabilidad de que la persona


a) experimente hipertensión, dado que es un fumador empedernido.

$H$ (Hipertenso)

$E$ (Fumador empedernido).

$P(H|E)$:

$$P(H|E) = \frac{\text{Hipertensos y Empedernidos}}{\text{Total de Empedernidos}}$$

$$P(H|E) = \frac{30}{49} \approx \mathbf{0.6122}$$

b) sea un no fumador, dado que no ha presentado problemas de hipertensión.

$NF$ (No fumador)

$NH$ (No hipertenso).

$P(NF|NH)$:

$$P(NF|NH) = \frac{\text{No fumadores y No hipertensos}}{\text{Total de No hipertensos}}$$

$$P(NF|NH) = \frac{48}{93} \approx \mathbf{0.5161}$$

####PREGUNTA 10
Problema de probabilidad condicional
En una fábrica se producen piezas mediante dos máquinas, A y B.

La máquina A produce el 60% del total de las piezas.
La máquina B produce el 40% restante.
Sin embargo, no son igual de precisas:

La máquina A produce 2% de piezas defectuosas.
La máquina B produce 5% de piezas defectuosas.
Un inspector toma una pieza al azar del inventario general y encuentra que la pieza está defectuosa.

Pregunta:
¿Cuál es la probabilidad de que la pieza haya sido fabricada por la máquina B?

* $P(A) = 0.60$: Probabilidad de que la pieza sea de la Máquina A.
* $P(B) = 0.40$: Probabilidad de que la pieza sea de la Máquina B.
* $P(D|A) = 0.02$: Probabilidad de defecto dado que es de la Máquina A.
* $P(D|B) = 0.05$: Probabilidad de defecto dado que es de la Máquina B.

Primero calculamos la probabilidad total de que una pieza sea defectuosa ($P(D)$):
$$P(D) = P(A) \cdot P(D|A) + P(B) \cdot P(D|B)$$

Calculamos la probabilidad de que la pieza provenga de la Máquina B dado que ya sabemos que es defectuosa ($P(B|D)$):
$$P(B|D) = \frac{P(B) \cdot P(D|B)}{P(D)}$$

```python
p_a = 0.60
p_b = 0.40
p_d_dado_a = 0.02
p_d_dado_b = 0.05

p_d = (p_a * p_d_dado_a) + (p_b * p_d_dado_b)
p_b_dado_d = (p_b * p_d_dado_b) / p_d

print(f"La probabilidad de que sea de la Máquina B es: {p_b_dado_d:.4f}")
```

##### Turning Data into Pictures
For this project, we've provided the following function to draw Mondrian paintings from data:


```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
def draw_mondrian( painting_id ):
    rects = features.query('painting_id == @painting_id')
    total_width = rects.eval("x + width").max()
    total_height = rects.eval("y + height").max()

    fig, ax = plt.subplots(figsize=(3, 3))

    for (idx, row) in rects.iterrows():
        x, y, w, h, rgb = row[['x','y','width','height','rgb']]
        patch = mpatches.Rectangle((x, y), w, h, facecolor=rgb)
        ax.add_patch(patch)

    ax.axis([0, total_width, 0, total_height])
    ax.set_aspect('equal')
    ax.axis('off')
    fig.text(0.5, 0.01, painting_id, ha="center", fontsize=14)
```

Internally, this function uses Matplotlib to draw each feature as a rectangular patch of color. While the details of this function are not the focus of this project, feel free to take a look at the code to see how it works.

```python
# 2. Diferencia en la altura del bolsillo delantero entre estilos (skinny vs straight) por género

# Agrupamos por género y estilo, y calculamos el promedio
promedio_estilo_genero = df.groupby(['gender', 'style'])['height_front'].mean().unstack()

print("--- 2. Altura promedio del bolsillo delantero por género y estilo (cm) ---")
display(promedio_estilo_genero)

# Calculamos las diferencias dentro del mismo género
diff_men = promedio_estilo_genero.loc['men', 'straight'] - promedio_estilo_genero.loc['men', 'skinny']
diff_women = promedio_estilo_genero.loc['women', 'straight'] - promedio_estilo_genero.loc['women', 'skinny']

print(f"\nEn hombres, el estilo straight es {diff_men:.2f} cm más profundo que el skinny.")
print(f"En mujeres, el estilo straight es {diff_women:.2f} cm más profundo que el skinny.")
```

```python
# 1. Conteo de territorios por el primer día de la semana

conteo_territorios = df_first_day['first_day'].value_counts()

print("--- 1. Número de territorios por primer día de la semana ---")
display(conteo_territorios.to_frame(name='Numero_de_Territorios'))
```

```python
# 2. Población total que comienza la semana cada día

# Hacemos un "merge" de los datos del primer día y la población usando el código del país ('alpha3')
df_merged_pop = pd.merge(df_first_day, df_pop, on='alpha3', how='inner')

# Agrupamos por el primer día y sumamos la población (en millones)
poblacion_por_dia = df_merged_pop.groupby('first_day')['population'].sum()

# Ordenamos de mayor a menor población
poblacion_por_dia = poblacion_por_dia.sort_values(ascending=False)

print("--- 2. Población (en millones) que comienza la semana cada día ---")
display(poblacion_por_dia.to_frame(name='Poblacion_Total_Millones'))
```

Primer intento de modelar los años de escolaridad y el ingreso estimado.   
Utilizar el primer punto y el útlimo para trazar una recta:   
$$  m = \frac{y_{2}-y_{1}}{x_{2}-x_{1}} $$
$$ b = y_{1} - m*x_{1} $$

```python
x1 = anios_escolaridad[0]
y1 = ingreso_mensual_estimado_mx[0]
x2 = anios_escolaridad[-1]
y2 = ingreso_mensual_estimado_mx[-1]

m = (y2 - y1) / (x2 - x1)
b = y1 - m * x1

print(f"La ecuación de la recta es: y = {m:.2f}x + {b:.2f}")
```

Valores predichos
: Son los ingresos mensuales que nuestra recta de regresión predice para cada año de escolaridad.

Errores (residuales): Es la diferencia entre el ingreso mensual real (ingreso_mensual_estimado_mx) y el ingreso mensual predicho (y_pred). Un error positivo significa que la predicción fue menor que el valor real, y un error negativo indica que la predicción fue mayor.

Errores al cuadrado: Hemos elevado al cuadrado cada uno de los errores. Esto se hace para eliminar los signos negativos (de modo que los errores positivos y negativos no se cancelen entre sí) y para penalizar más los errores grandes.

Suma de errores cuadrados: Es la suma de todos los errores al cuadrado. Este valor nos da una medida total de la 'distancia' de nuestros puntos a la línea.

Error Cuadrático Medio (MSE): Es la suma de los errores cuadrados dividida por el número total de puntos de datos. El MSE es una medida promedio de la magnitud de los errores. Un MSE más bajo indica que el modelo se ajusta mejor a los datos.

En este caso, el Error Cuadrático Medio (MSE) es de aproximadamente 13,196,667.31. Este número nos da una idea del rendimiento general de la recta que trazamos, mostrando la magnitud promedio de los errores al cuadrado entre los valores observados y los valores predichos por la línea.

Fórmula del MSE (Error Cuadrático Medio) del Inglés Mean Squared Error

$$ MSE = \frac{\Sigma_{i=1}^{n}(Y_{i}-\hat{Y_{i}})^{2}}{n}$$

$MSE = $ error cuadrático medio   
$n = $ número de datos   
$Y_{i} = $ valores observados   
$\hat{Y_{i}} = $ valores predichos   


```python
# 1. Calcular los valores predichos (y_pred) usando la ecuación de la recta
y_pred = m * anios_escolaridad + b
y_pred
```

```python
# 2. Calcular los errores (residuales): la diferencia entre los valores reales y los predichos
errores = ingreso_mensual_estimado_mx - y_pred
errores
```

```python
# 3. Elevar al cuadrado cada error
errores_cuadrados = errores**2
errores_cuadrados
```

```python
# 4. Sumar los errores cuadrados
suma_errores_cuadrados = np.sum(errores_cuadrados)
suma_errores_cuadrados
```

```python
# 5. Dividir la suma de los errores cuadrados por el número de puntos para obtener el MSE
mse = suma_errores_cuadrados / len(anios_escolaridad)
mse
```

```python
#@title Simulación Monte Carlo de P(H|E)
import numpy as np
import matplotlib.pyplot as plt

# Datos del problema
#            Empleado  Desempleado  Total
# Hombre        460         40       500
# Mujer         140         260      400
# Total         600         300      900

# Probabilidad analítica P(H|E) = P(H ∩ E) / P(E)
P_H_and_E = 460/900
P_E = 600/900
P_H_given_E_exact = P_H_and_E / P_E

# Simulación Monte Carlo
N = 300_000  # número de experimentos
rng = np.random.default_rng(42)

# Estados: 0=(H, E), 1=(H, D), 2=(M, E), 3=(M, D)
counts = np.array([460, 40, 140, 260], dtype=int)
probs = counts / counts.sum()

# Simulamos elegir un individuo al azar N veces
samples = rng.choice(4, size=N, p=probs)

# Evento E: empleado -> estados {0, 2}
es_empleado = np.isin(samples, [0, 2])
# Evento H ∩ E: estado 0
es_hombre_y_empleado = (samples == 0)

# Estimaciones Monte Carlo
P_H_and_E_hat = es_hombre_y_empleado.mean()
P_E_hat = es_empleado.mean()
P_H_given_E_hat = P_H_and_E_hat / P_E_hat

print(f"Valor exacto  P(H|E)     = {P_H_given_E_exact:.6f}")
print(f"Estimación MC P(H|E)     = {P_H_given_E_hat:.6f}")
print(f"P(H∩E) (MC)              = {P_H_and_E_hat:.6f}")
print(f"P(E) (MC)                = {P_E_hat:.6f}")

# Curva de convergencia de P(H|E) por prefijos
cum_H_and_E = np.cumsum(es_hombre_y_empleado)
cum_E = np.cumsum(es_empleado)
valid = cum_E > 0
P_curve = np.empty(N)
P_curve[:] = np.nan
P_curve[valid] = cum_H_and_E[valid] / cum_E[valid]

# Graficamos una muestra espaciada de puntos para eficiencia visual
idx = np.linspace(1, N, 4000, dtype=int)

plt.figure(figsize=(8, 4.8))
plt.plot(idx, P_curve[idx-1], lw=1, label='Estimación simulada')
plt.axhline(P_H_given_E_exact, color='crimson', linestyle='--',
            label=f"Valor exacto = {P_H_given_E_exact:.6f}")
plt.title('Convergencia de la simulación de $P(H\\,|\\,E)$')
plt.xlabel('Número de simulaciones')
plt.ylabel('Probabilidad estimada')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
#@title Simulación Monte Carlo de P(H|E)
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Datos del problema
#            Empleado  Desempleado  Total
# Hombre        460         40       500
# Mujer         140         260      400
# Total         600         300      900
```

```python
# Simulación Monte Carlo
N = 25_000  # número de experimentos
rng = np.random.default_rng(42)
rng
```

```python
# Estados: 0=(H, E), 1=(H, D), 2=(M, E), 3=(M, D)
counts = np.array([460, 40, 140, 260], dtype=int)
probs = counts / counts.sum()
probs
```

```python
# Simulamos elegir un individuo al azar N veces
samples = rng.choice(4, size=N, p=probs)
samples
```

```python
# Evento E: empleado -> estados {0, 2}
es_empleado = np.isin(samples, [0, 2])
es_empleado
```

```python
# Evento H ∩ E: estado 0
es_hombre_y_empleado = (samples == 0)
es_hombre_y_empleado
```

```python
# Probabilidad analítica P(H|E) = P(H ∩ E) / P(E)
P_H_and_E = 460/900
P_E = 600/900
P_H_given_E_exact = P_H_and_E / P_E
P_H_given_E_exact
```

```python
# Estimaciones Monte Carlo
P_H_and_E_hat = es_hombre_y_empleado.mean()
P_E_hat = es_empleado.mean()
P_H_given_E_hat = P_H_and_E_hat / P_E_hat

print(f"Valor exacto  P(H|E)     = {P_H_given_E_exact:.6f}")
print(f"Estimación MC P(H|E)     = {P_H_given_E_hat:.6f}")
print(f"P(H∩E) (MC)              = {P_H_and_E_hat:.6f}")
print(f"P(E) (MC)                = {P_E_hat:.6f}")
```

```python
# Curva de convergencia de P(H|E) por prefijos
cum_H_and_E = np.cumsum(es_hombre_y_empleado)
cum_E = np.cumsum(es_empleado)
valid = cum_E > 0
P_curve = np.empty(N)
P_curve[:] = np.nan
P_curve[valid] = cum_H_and_E[valid] / cum_E[valid]

# Graficamos una muestra espaciada de puntos para eficiencia visual
idx = np.linspace(1, N, 4000, dtype=int)

plt.figure(figsize=(8, 4.8))
plt.plot(idx, P_curve[idx-1], lw=1, label='Estimación simulada')
plt.axhline(P_H_given_E_exact, color='crimson', linestyle='--',
            label=f"Valor exacto = {P_H_given_E_exact:.6f}")
plt.title('Convergencia de la simulación de $P(H\\,|\\,E)$')
plt.xlabel('Número de simulaciones')
plt.ylabel('Probabilidad estimada')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

###K‑Nearest Neighbors (K-NN) k vecinos más cercanos

####🟢 Ejercicio 1: k‑NN manual (k = 1, distancia euclidiana)

#####Datos

```python
import math

# Datos de entrenamiento
X = [(2,4), (4,4), (4,2), (6,2)]
y = ['Rojo', 'Rojo', 'Azul', 'Azul']

# Punto a clasificar
P = (3, 3)
```

#####Función de distancia euclidiana

```python
def distancia_euclidiana(p1, p2):
    return math.sqrt((p1[0]-p2[0])**2 + (p1[1]-p2[1])**2)
```

#####Cálculo del vecino más cercano

```python
distancias = []
for i in range(len(X)):
    d = distancia_euclidiana(P, X[i])
    distancias.append((d, y[i]))

# Ordenar por distancia
distancias.sort(key=lambda x: x[0])

print(distancias)
print("Clase asignada:", distancias[0][1])
```

✅ Objetivo didáctico: entender cómo funciona k‑NN desde cero.

####🟡 Ejercicio 2: k‑NN manual con k = 3

```python
from collections import Counter

k = 3
vecinos = distancias[:k]

clases = [clase for _, clase in vecinos]
resultado = Counter(clases).most_common(1)

print("Vecinos:", vecinos)
print("Clase asignada:", resultado[0][0])
```

✅ Tema clave: votación por mayoría.

#####📊 Dataset: Clientes para un crédito
Supón que tenemos datos de clientes y queremos clasificar a un nuevo cliente según su riesgo crediticio.

| Cliente | Edad (años) | Ingreso mensual ($) | Riesgo |

📌 Observa:

Edad: valores ~25–50

Ingreso: valores ~8,000–60,000

👉 El ingreso domina la distancia si no se normaliza.

#####🎯 Punto a clasificar
Nuevo cliente:

Edad = 40

Ingreso = 20,000

#####🟥 Parte A: k‑NN SIN normalización

```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier

# Dataset
X = np.array([
[25, 8000],
[30, 12000],
[45, 45000],
[50, 60000],
[35, 15000],
[48, 55000]
])

y = np.array(['Alto', 'Alto', 'Bajo', 'Bajo', 'Alto', 'Bajo'])

nuevo_cliente = np.array([[40, 20000]])

# Modelo k-NN
modelo = KNeighborsClassifier(n_neighbors=3)
modelo.fit(X, y)

# Predicción
modelo.predict(nuevo_cliente)
```

🔎 ¿Qué está pasando aquí?

La distancia está dominada casi totalmente por el ingreso

La edad tiene muy poco impacto

El modelo buscará clientes con ingresos cercanos a 20,000


✅ Resultado típico:
Riesgo = Alto



📌 Interpretación

El algoritmo ignora parcialmente la edad, aunque esta podría ser relevante para el riesgo.

#####🟢 Parte B: k‑NN CON normalización

Ahora normalizamos usando StandardScaler.

```python
from sklearn.preprocessing import StandardScaler

# Escalamiento
scaler = StandardScaler()
X_norm = scaler.fit_transform(X)
nuevo_norm = scaler.transform(nuevo_cliente)

# Modelo k-NN
modelo_norm = KNeighborsClassifier(n_neighbors=3)
modelo_norm.fit(X_norm, y)

# Predicción
modelo_norm.predict(nuevo_norm)
```

🔎 ¿Qué cambia ahora?

Edad e ingreso tienen:

media = 0

desviación estándar = 1

Ambas variables pesan por igual en la distancia


✅ Resultado frecuente:

Riesgo = Bajo (puede cambiar dependiendo de k)

#####¿Qué es el "Método del Codo"?

El método del codo es una técnica visual que se utiliza para determinar el número óptimo de vecinos ($k$) en el algoritmo K-NN (también se usa mucho en K-Means).Funciona de la siguiente manera:

1. Ejecutas el algoritmo K-NN varias veces, iterando sobre diferentes valores de $k$ (por ejemplo, del 1 al 20).

2. Para cada valor de $k$, calculas la tasa de error (qué porcentaje de las predicciones fueron incorrectas en tu conjunto de prueba).

3. Graficas estos valores: el eje X representa la $k$ y el eje Y representa la tasa de error.

4. Buscas el "codo" en la gráfica: el punto donde la tasa de error cae drásticamente y luego comienza a estabilizarse o a subir de nuevo. Ese punto de inflexión te indica el valor de $k$ que ofrece el mejor equilibrio: ni tan bajo como para sobreajustar el modelo (overfitting), ni tan alto como para perder precisión (underfitting).

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, accuracy_score

iris = load_iris()
X = iris.data    #largo/ancho de sépalo y pétalo
y = iris.target  # 3 especies de flores

#80% entrenamiento, 20% prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# 3. Investigar el "Codo" para encontrar el mejor valor de k
tasas_error = []

#k desde 1 hasta 20
rango_k = range(1, 21)
for i in rango_k:
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train, y_train)
    pred_i = knn.predict(X_test)

    tasas_error.append(np.mean(pred_i != y_test))

plt.figure(figsize=(10, 6))
plt.plot(rango_k, tasas_error, color='blue', linestyle='dashed', marker='o',
         markerfacecolor='red', markersize=8)
plt.title('Tasa de Error vs. Valor de K (Método del Codo)')
plt.xlabel('Valor de K')
plt.ylabel('Tasa de Error')
plt.xticks(rango_k)
plt.grid(True)
plt.show()

k_optimo = 3

print(f"\nEntrenando el modelo final con K = {k_optimo}...\n")
knn_optimo = KNeighborsClassifier(n_neighbors=k_optimo)
knn_optimo.fit(X_train, y_train)
predicciones = knn_optimo.predict(X_test)

# Resultados
print(f"Precisión (Accuracy): {accuracy_score(y_test, predicciones) * 100:.2f}%")
print("\n--- Reporte de Clasificación ---")
print(classification_report(y_test, predicciones, target_names=iris.target_names))
```
