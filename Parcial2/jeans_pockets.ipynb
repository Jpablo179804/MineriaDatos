# Jean pockets

Are women jean pockets ridiculously small? Let's look at the data!

The file `jean-pocket-measurements.csv` shows pocket measurements for 20 popular brands.
Four pairs of jeans from each brand were measured: men's and women's skinny and straight styles. All jeans were designated a 32-inch waistband.

The following measurements are included:
- `height_front`: maximum front pocket height (cm)
- `height_back`: maximum back pocket height (cm)
- `width_front`: maximum front pocket width (cm)
- `width_back`: maximum back pocket width (cm)

```python
import pandas as pd
df = pd.read_csv('jean-pocket-measurements.csv')
df.head()
```

### Project Ideas

- Is there a significant difference in pocket `height_front` between skinny and straight styles within the same gender?

¿Cuál es la diferencia promedio en la altura del bolsillo delantero (`height_front`) entre pantalones de mujeres y hombres?
¿Hay una diferencia significativa en la altura del bolsillo delantero entre los estilos *skinny* y *straight* dentro del mismo género?
¿Cómo se comparan los tamaños de los bolsillos traseros entre pantalones de mujeres y hombres?
¿Qué porcentaje de los pantalones de mujeres y hombres pueden acomodar cómodamente un teléfono promedio en los bolsillos delanteros?

```python
# 1. Diferencia promedio en la altura del bolsillo delantero (height_front) entre mujeres y hombres

# Agrupamos por género y calculamos el promedio de la altura del bolsillo delantero
promedio_frente_por_genero = df.groupby('gender')['height_front'].mean()

print("--- 1. Promedio de altura del bolsillo delantero (cm) ---")
display(promedio_frente_por_genero)

diferencia = promedio_frente_por_genero['men'] - promedio_frente_por_genero['women']
print(f"\nLos bolsillos delanteros de los hombres son en promedio {diferencia:.2f} cm más altos que los de las mujeres.")
```

```python
# 2. Diferencia en la altura del bolsillo delantero entre estilos (skinny vs straight) por género

# Agrupamos por género y estilo, y calculamos el promedio
promedio_estilo_genero = df.groupby(['gender', 'style'])['height_front'].mean().unstack()

print("--- 2. Altura promedio del bolsillo delantero por género y estilo (cm) ---")
display(promedio_estilo_genero)

# Calculamos las diferencias dentro del mismo género
diff_men = promedio_estilo_genero.loc['men', 'straight'] - promedio_estilo_genero.loc['men', 'skinny']
diff_women = promedio_estilo_genero.loc['women', 'straight'] - promedio_estilo_genero.loc['women', 'skinny']

print(f"\nEn hombres, el estilo straight es {diff_men:.2f} cm más profundo que el skinny.")
print(f"En mujeres, el estilo straight es {diff_women:.2f} cm más profundo que el skinny.")
```

```python
# 3. Comparación de los bolsillos traseros entre mujeres y hombres

# Calculamos el promedio de la altura y anchura del bolsillo trasero por género
promedio_trasero = df.groupby('gender')[['height_back', 'width_back']].mean()

print("--- 3. Tamaño promedio del bolsillo trasero (cm) ---")
display(promedio_trasero)

diff_height_back = promedio_trasero.loc['men', 'height_back'] - promedio_trasero.loc['women', 'height_back']
diff_width_back = promedio_trasero.loc['men', 'width_back'] - promedio_trasero.loc['women', 'width_back']

print(f"\nLos bolsillos traseros de los hombres son en promedio {diff_height_back:.2f} cm más altos y {diff_width_back:.2f} cm más anchos que los de las mujeres.")
```

```python
# 4. ¿Qué porcentaje de pantalones pueden acomodar un teléfono en el bolsillo delantero?

# Asumiremos la altura de un teléfono estándar actual (aprox. 15.0 cm). Puedes cambiar este valor por el de tu teléfono real.
altura_telefono = 15.0

# Filtramos por género
df_women = df[df['gender'] == 'women']
df_men = df[df['gender'] == 'men']

# Calculamos el porcentaje donde el bolsillo delantero es mayor o igual a la altura del teléfono
pct_mujeres_cabe = (df_women['height_front'] >= altura_telefono).mean() * 100
pct_hombres_cabe = (df_men['height_front'] >= altura_telefono).mean() * 100

print(f"--- 4. Prueba de ajuste para un teléfono de {altura_telefono} cm de alto ---")
print(f"Porcentaje de pantalones de mujeres donde cabe el teléfono: {pct_mujeres_cabe:.2f}%")
print(f"Porcentaje de pantalones de hombres donde cabe el teléfono: {pct_hombres_cabe:.2f}%")
```
