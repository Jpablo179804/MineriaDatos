Lee ambos CSVs en dfc (clientes) y dfo (órdenes).

ordenes.csv

Realiza un left merge de dfc con dfo por cliente_id, manteniendo todas las filas de dfc.
Calcula:
(a) Número total de filas del resultado del left merge.
(b) Monto total por cliente (suma de monto) y reemplaza NaN por 0.
Pregunta numérica:

¿Cuál es el monto total por cliente cliente_id=1?

```python
import pandas as pd
import io
from google.colab import files

print("Sube los archivos clientes.csv y ordenes.csv:")
uploaded = files.upload()

dfc = pd.read_csv(io.BytesIO(uploaded['clientes.csv']))
dfo = pd.read_csv(io.BytesIO(uploaded['ordenes.csv']))

df_merged = pd.merge(dfc, dfo, on='cliente_id', how='left')

total_filas = len(df_merged)
print(f"\n(a) Número total de filas del left merge: {total_filas}")

monto_por_cliente = df_merged.groupby('cliente_id')['monto'].sum().fillna(0)

print("\n(b) Monto total por cliente:")
print(monto_por_cliente)

monto_cliente_1 = monto_por_cliente.get(1, 0)
print(f"\n---> Respuesta Numérica: El monto total para el cliente_id=1 es: {monto_cliente_1}")
```

Carga el CSV en df, selecciona solo columnas: ts, temperatura, humedad.
Convierte ts a datetime manejando formatos mixtos (usa format='mixed' y errors='coerce').
Reemplaza valores faltantes en temperatura con la mediana y en humedad con la media (usa fillna).
Reporta:
(a) ¿Cuántos NaT quedaron en ts?
(b) Valor final de la mediana de temperatura tras imputación (redondea a 1 decimal).
(c) Valor final de la media de humedad tras imputación (redondea a 3 decimales).
Formato de entrega: tres números (en orden a, b, c):
Enter answer

Enter answer

Enter answer


```python
import pandas as pd
import io
from google.colab import files

print("Sube el archivo sensores.csv:")
uploaded = files.upload()

df = pd.read_csv(io.BytesIO(uploaded['sensores.csv']))
df = df[['ts', 'temperatura', 'humedad']]

df['ts'] = pd.to_datetime(df['ts'], format='mixed', errors='coerce')

df['temperatura'] = pd.to_numeric(df['temperatura'], errors='coerce')
df['humedad'] = pd.to_numeric(df['humedad'], errors='coerce')

mediana_temp = df['temperatura'].median()
media_hum = df['humedad'].mean()

df['temperatura'] = df['temperatura'].fillna(mediana_temp)
df['humedad'] = df['humedad'].fillna(media_hum)

nat_ts = df['ts'].isna().sum()

mediana_final_temp = round(df['temperatura'].median(), 1)

media_final_hum = round(df['humedad'].mean(), 3)

print(f"\n(a) NaT en ts: {nat_ts}")
print(f"(b) Mediana final de temperatura: {mediana_final_temp}")
print(f"(c) Media final de humedad: {media_final_hum}")
print(f"\nRespuesta final: {nat_ts}, {mediana_final_temp}, {media_final_hum}")
```

Completa el siguiente fragmento de código para entrenar un clasificador Multinomial Naive Bayes con datos de texto usando CountVectorizer. Llena los espacios marcados con ____ para que el código funcione correctamente.

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

textos = [
   "el gato duerme",
   "el perro ladra",
   "el gato maulla",
   "el perro corre"
]

y = ["gato", "perro", "gato", "perro"]

modelo = Pipeline([
   ("vect", CountVectorizer()),
   ("clf", MultinomialNB())
])

modelo.fit(textos, y)

pred = modelo.predict(["el gato corre"])
print(f"Predicción: {pred[0]}")
```

Selecciona el fragmento de código que correctamente entrena un clasificador de Spam vs. No‑Spam utilizando un vectorizador de texto y un algoritmo supervisado de scikit‑learn.

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline

emails = [
    "texto",
    "texto",
    "texto",
    "texto"
]
labels = [1, 0, 1, 0]

modelo = Pipeline([
    ("vect", CountVectorizer()),
    ("clf", MultinomialNB())
])

modelo.fit(emails, labels)

nuevo_correo = ["Oferta exclusiva, haz clic aquí"]
prediccion = modelo.predict(nuevo_correo)
print("Spam" if prediccion[0] == 1 else "No-Spam")
```
