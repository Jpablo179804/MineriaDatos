```python
# Importamos la librería y cargamos los tres archivos CSV necesarios
import pandas as pd

df_first_day = pd.read_csv('first-day-of-week.csv')
df_pop = pd.read_csv('population.csv')
df_regions = pd.read_csv('four-regions.csv')

print("¡Datos cargados correctamente!")
```

```python
# 1. Conteo de territorios por el primer día de la semana

conteo_territorios = df_first_day['first_day'].value_counts()

print("--- 1. Número de territorios por primer día de la semana ---")
display(conteo_territorios.to_frame(name='Numero_de_Territorios'))
```

```python
# 2. Población total que comienza la semana cada día

# Hacemos un "merge" de los datos del primer día y la población usando el código del país ('alpha3')
df_merged_pop = pd.merge(df_first_day, df_pop, on='alpha3', how='inner')

# Agrupamos por el primer día y sumamos la población (en millones)
poblacion_por_dia = df_merged_pop.groupby('first_day')['population'].sum()

# Ordenamos de mayor a menor población
poblacion_por_dia = poblacion_por_dia.sort_values(ascending=False)

print("--- 2. Población (en millones) que comienza la semana cada día ---")
display(poblacion_por_dia.to_frame(name='Poblacion_Total_Millones'))
```

```python
# 3. Regiones que comienzan en domingo, lunes, etc. y su división

# Hacemos un "merge" de los datos del primer día y las regiones usando 'alpha3'
df_merged_reg = pd.merge(df_first_day, df_regions, on='alpha3', how='inner')

# Usamos crosstab para crear una tabla cruzada que cuente cuántos territorios por región usan cada día
region_vs_dia = pd.crosstab(df_merged_reg['four_regions'], df_merged_reg['first_day'])

print("--- 3. Distribución del primer día de la semana por región (Cantidad de territorios) ---")
display(region_vs_dia)

print("\nAnálisis rápido:")
print("- Observa qué región tiene el número más alto en la columna 'sun' (domingo).")
print("- Observa qué región domina ampliamente en la columna 'mon' (lunes).")
print("- Busca qué continente tiene números más parecidos entre varios días para ver la división.")
```
