#Liga MX jornada 17


```python
from google.colab import files
import pandas as pd
from sklearn.ensemble import RandomForestClassifier

print("Sube tu archivo 'dataset_liga_mx_jornadas_13_14_15_base.csv':")
uploaded = files.upload()
nombre_archivo = list(uploaded.keys())[0]

df = pd.read_csv(nombre_archivo)
features = ['diff_puntos', 'diff_racha', 'diff_posicion']
X = df[features]
y = df['resultado']

rf = RandomForestClassifier(n_estimators=300, max_depth=5, random_state=42)
rf.fit(X, y)
print("\nmodelo entrenado")
```

```python
#[diff_puntos, diff_racha, diff_posicion]
jornada_17_data = {
    'Puebla vs Querétaro': [-2, 1, 2],
    'Pachuca vs Pumas': [1, 2, -1],
    'Tigres vs Mazatlán': [10, 5, -9],
    'Toluca vs León': [5, -4, -1],
    'Chivas vs Tijuana': [18, 12, -9],
    'América vs Atlas': [-1, -8, 2],
    'FC Juárez vs Atlético San Luis': [-2, -9, 4],
    'Santos vs Monterrey': [-7, -2, 5],
    'Cruz Azul vs Necaxa': [15, 8, -8]
}

clases = {0: "Gana visitante", 1: "Empatan", 2: "Gana local"}

print("Predicciones Jornada 17:\n" + "-"*35)
for partido, caracteristicas in jornada_17_data.items():
    X_pred = pd.DataFrame([caracteristicas], columns=features)
    prediccion = rf.predict(X_pred)[0]
    print(f"{partido}:\n--> {clases[prediccion]}\n")
```
